```python
import pandas as pd
import numpy as np
import networkx as nx
import os
import matplotlib.pyplot as plt
from pyvis.network import Network
import seaborn as sns
import pyvis 
```

```python
relationship_df = pd.read_csv('country_relationship.csv')
```

```python
relationship_df.head()
```

```python
G = nx.from_pandas_edgelist(relationship_df,
                                               source = "source",
                target = "target",
                edge_attr =  "value",
                create_using = nx.Graph())
```

```python
plt.figure(figsize = (10,10))
pos = nx.kamada_kawai_layout (G)
nx.draw(G, with_labels = True, node_color = 'red', edge_cmap = plt.cm.Blues, pos = pos)
```

```python
plt.figure(figsize = (10,10))
pos = nx.kamada_kawai_layout (G)
nx.draw(G, with_labels = True, node_color = 'red', edge_cmap = plt.cm.Blues, pos = pos)
```

```python
# Define net object

net = Network()
net = Network(notebook = True,width="900px", height="800px", bgcolor='#222222', font_color='white')
```

```python
# Define nodes from the G object
node_degree = dict(G.degree)
```

```python
# Setting up node size attributes
nx.set_node_attributes(G, node_degree, 'size')
net.from_nx(G)
net.show_buttons(filter_= True)
net.repulsion()
net.show("countries.html")
```

```python
from cdlib import algorithms
coms = algorithms.leiden(G)
```

```python
new_com = coms.to_node_community_map()
```

```python
new_com
```

```python
# Put the output from new_com into a dictionary

dict_com = {k:v[0] for k, v in new_com.items()}
```

```python
# Rerun network graph with the communities as an argument

nx.set_node_attributes(G, dict_com, 'group')
com_net = Network(notebook = True, width="1300px", height="700px", bgcolor='#222222', font_color='white')
com_net.from_nx(G)
com_net.show_buttons(filter_ = True)
com_net.repulsion()
com_net.show("countries_leiden.html")
```

```python
# Degree centrality

degree_dict = nx.degree_centrality(G)
degree_dict
```

```python
degree_df = pd.DataFrame(degree_dict.items(), columns = ['countries','centrality'] )
degree_df = degree_df.sort_values(by = 'centrality', ascending = False)
degree_df.head()
```

```python
#degree_df.sort_values(by = 'centrality')
```

```python
plt.figure(figsize = (10, 10))
with sns.dark_palette("xkcd:blue", 10):
      sns.barplot(x = 'centrality', y = 'countries',
     saturation = 0.9, data=degree_df).set_title("node closeness by country : degree centrality")
```

```python
# closeness centrality

close_dict = nx.closeness_centrality(G)
close_dict
```

```python
close_df = pd.DataFrame(close_dict.items(), columns = ['countries','centrality'] )
close_df = close_df.sort_values(by = 'centrality', ascending = False)
close_df.head()
```

```python
plt.figure(figsize = (10, 10))
with sns.dark_palette("xkcd:blue", 10):
      sns.barplot(x = 'centrality', y = 'countries',
     saturation = 0.9, data=close_df).set_title("node closeness by country : closeness centrality")
```

```python
# betweenness centrality

between_dict = nx.betweenness_centrality(G)
between_dict
```

```python
between_df = pd.DataFrame(between_dict.items(), columns = ['countries','centrality'] )
between_df = between_df.sort_values(by = 'centrality', ascending = False)
between_df.head()
```

```python
plt.figure(figsize = (10, 10))
with sns.dark_palette("xkcd:blue", 10):
      sns.barplot(x = 'centrality', y = 'countries',
     saturation = 0.9, data=between_df).set_title("node closeness by country : betweenness centrality")
```
